## a way to delete features

just select the data between lower and upper bound

```python
from sklearn.datasets import make_blobs
```

```python
import numpy as np
```

```python
features, *_ = make_blobs(n_samples = 10,
                         n_features = 2,
                         centers = 1,
                         random_state = 1)

feature = features[:, 0]

def indices_of_outliers(x:int) -> np.array(int):
  q1, q3 = np.percentile(x, [25, 75])
  iqr = q3 - q1
  lower, upper = q1-1.5*iqr, q3+1.5*iqr
  return np.where((x > upper) | (x < lower))

indices_of_outliers(feature)


```
